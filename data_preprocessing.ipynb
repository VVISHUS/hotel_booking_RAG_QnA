# Data Duplication Analysis

```python

def analyze_duplicates(file_path=None, df=None):

    if file_path:
        print(f"Loading data from {file_path}...")
        try:
            df = pd.read_csv(file_path)
        except Exception as e:
            print(f"Error loading file: {e}")
            return None
    elif df is None:
        print("No data provided. Using sample data for demonstration.")
        return None
        
    print(f"Dataset shape: {df.shape}")
    
    # 1. Check for exact duplicates
    exact_dupes = df.duplicated().sum()
    print(f"\n1. Exact duplicates: {exact_dupes} rows ({exact_dupes/len(df)*100:.2f}% of data)")
    
    # 2. Analyze specific subset duplicates to understand business context
    # These are combinations that might represent legitimate duplicates or data entry errors
    
    # 2.1 Possible business-significant duplication sets
    duplication_sets = [
        # Same guest, same arrival, same room type (potential duplicate booking)
        ['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 
         'adults', 'children', 'babies', 'country', 'reserved_room_type'],
        
        # Same booking but different handling (might be system duplicates)
        ['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month',
         'adults', 'children', 'lead_time', 'country', 'adr'],
        
        # Same hotel, date, room - could be coincidence or system duplicate
        ['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 
         'reserved_room_type', 'assigned_room_type', 'adr'],
        
        # Potential technical identifiers
        ['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 
         'agent', 'company', 'adults', 'children']
    ]
    
    duplicate_analysis = {}
    for i, cols in enumerate(duplication_sets, 1):
        set_name = f"Set {i}"
        dupes = df.duplicated(subset=cols, keep=False)
        num_dupes = dupes.sum()
        duplicate_analysis[set_name] = {
            'columns': cols,
            'duplicate_count': num_dupes,
            'percentage': num_dupes/len(df)*100
        }
        print(f"\n2.{i} Duplicates by {set_name} ({', '.join(cols)}):")
        print(f"   Found {num_dupes} rows ({num_dupes/len(df)*100:.2f}% of data)")
    
    # 3. Check for near-duplicates with small differences
    print("\n4. Analyzing potential near-duplicates:")
    
    # Find some rows that might be duplicates except for small differences
    near_dupes_cols = ['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 
                       'adults', 'children', 'reserved_room_type', 'adr', 'country']
    
    # Group by the near dupes columns
    near_dupes = df.groupby(near_dupes_cols).filter(lambda x: len(x) > 1)
    print(f"   Found {len(near_dupes)} potential near-duplicate rows")
    
    # 4.. Check for varying fields in duplicates
    if len(near_dupes) > 0:
        print("\n5. Fields that commonly vary in near-duplicates:")
        
        # Group by business key and check which fields vary within groups
        vary_fields = {}
        for field in df.columns:
            if field not in near_dupes_cols:
                # For each potential group of dupes, check if this field varies
                field_varies = near_dupes.groupby(near_dupes_cols)[field].nunique().mean()
                vary_fields[field] = field_varies
        
        # Sort by most varying fields
        sorted_vary = sorted(vary_fields.items(), key=lambda x: x[1], reverse=True)
        for field, variation in sorted_vary[:10]:  # Top 10 varying fields
            print(f"   {field}: varies in {variation:.2f} of near-duplicate groups")
    
    # 5. Visualize duplicate patterns
    
    dupe_analysis_df = pd.DataFrame(duplicate_analysis).T
    dupe_analysis_df['set_name'] = dupe_analysis_df.index
    
    
    # Combined visualization
    fig = make_subplots(
        rows=2, cols=1,
        subplot_titles=["Duplicate Percentages by Different Column Sets", 
                        "Distribution of Duplicate Group Sizes"]
    )
    
    # Add bars for duplicate percentages
    fig.add_trace(
        go.Bar(
            x=dupe_analysis_df['set_name'],
            y=dupe_analysis_df['percentage'],
            text=[f"{p:.2f}%" for p in dupe_analysis_df['percentage']],
            textposition="outside",
            name="Duplicate %"
        ),
        row=1, col=1
    )
    
    fig.update_yaxes(title_text="Percentage of Dataset (%)", row=1, col=1)
    fig.update_xaxes(title_text="Column Set", row=1, col=1)
    
    return {
        'dataframe': df,
        'exact_dupes': exact_dupes,
        'duplicate_analysis': duplicate_analysis,
        'fig': fig
    }
```

```python
def analyze_revenue_trends(df):
    """Analyze revenue trends over time"""
    
    # Group by arrival month and calculate total revenue
    df['year_month'] = df['arrival_date'].dt.to_period('M')
    revenue_trends = df.groupby('year_month').agg({
        'revenue': 'sum',
        'is_canceled': 'count' 
    }).reset_index()
    
    revenue_trends['year_month'] = revenue_trends['year_month'].astype(str)
    revenue_trends = revenue_trends.rename(columns={'is_canceled': 'total_bookings'})
    
    # Create plot with dual y-axis
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    
    # Add revenue line
    fig.add_trace(
        go.Scatter(
            x=revenue_trends['year_month'],
            y=revenue_trends['revenue'],
            name="Total Revenue",
            line=dict(color="blue"),
            mode="lines+markers"
        ),
        secondary_y=False
    )
    
    # Add bookings line
    fig.add_trace(
        go.Scatter(
            x=revenue_trends['year_month'],
            y=revenue_trends['total_bookings'],
            name="Total Bookings",
            line=dict(color="green", dash="dot"),
            mode="lines+markers"
        ),
        secondary_y=True
    )
    
    fig.update_layout(
        title="Revenue and Bookings Trends Over Time",
        xaxis_title="Month",
        legend=dict(x=0.01, y=0.99, bgcolor="rgba(255,255,255,0.8)"),
        template="plotly_white"
    )
    
    fig.update_yaxes(title_text="Revenue (€)", secondary_y=False)
    fig.update_yaxes(title_text="Number of Bookings", secondary_y=True)
    
    fig.show()
    return fig
fig=analyze_revenue_trends(df)
```
