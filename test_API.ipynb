```python
import requests
```

```python
q="For which reason or faactor do people often cancel booking?"
```

```python

BASE_URL = "http://127.0.0.1:5000"

q1="What is the average booking rate for city hotels?"
def test_health_check():
    response = requests.get(f"{BASE_URL}/health_check")
    print("Health Check Response:", response.status_code)
    print("Response Body:", response.text)

def test_generate_result():
    payload = {
        "req_id": "test-001",
        "query": q,
        "descriptive": False

    }

    response = requests.post(f"{BASE_URL}/ask", json=payload)
    print("Generate Result Response:", response.status_code)
    try:
        print("Response JSON:", response.json())
    except:
        print("Response Text:", response.text)
```

```python

test_health_check()
test_generate_result()
```
